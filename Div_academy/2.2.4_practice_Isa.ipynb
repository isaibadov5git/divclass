```python
df = pd.DataFrame({
    'Team': ['A', 'A', 'B', 'B', 'C', 'C'],
    'Points': [10, 15, 10, 20, 10, 5],
    'Assists': [5, 7, 5, 10, 5, 2]
})
df
```

```python
df.groupby('Team').agg({
    'Points' : 'sum',
    'Assists' : 'sum'
})
df.rename
```

```python
df.groupby('Team').filter(lambda x: x["Points"].mean() > 12)
```

```python
def funct(x):
    return x - x.mean()

df['Points_Standardized'] = df.groupby('Team')['Points'].transform(funct)
df
```

```python
def ope(x):
    if x['Assists'].sum() > 10:
        return x['Points'] * 2
    else:
        return x['Points']
df.groupby('Team').apply(ope)
```
