```python
import pandas as pd

df = pd.read_csv(Path('/home/admin123/Documents/SAP/Div_academy/data/hotel_bookings.csv'))

df = df.replace('', pd.NA)

numeric_cols = ['lead_time', 'arrival_date_year', 'arrival_date_week_number',
                'arrival_date_day_of_month', 'stays_in_weekend_nights',
                'stays_in_week_nights', 'adults', 'children', 'babies',
                'previous_cancellations', 'previous_bookings_not_canceled',
                'booking_changes', 'days_in_waiting_list',
                'adr', 'required_car_parking_spaces',
                'total_of_special_requests']

df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric)

df['agent'] = pd.to_numeric(df['agent']).astype('Int64')
df['company'] = pd.to_numeric(df['company']).astype('Int64')


df['arrival_date_month'] = pd.Categorical(df['arrival_date_month'],
    categories=['January','February','March','April','May','June','July','August',
                'September','October','November','December'],
    ordered=True
)

df['arrival_month_num'] = df['arrival_date_month'].cat.codes + 1

df['reservation_status_date'] = pd.to_datetime(df['reservation_status_date'],
                                               )

cat_cols = [
    'hotel','meal','country','market_segment','distribution_channel',
    'reserved_room_type','assigned_room_type','deposit_type',
    'customer_type','reservation_status'
]

for c in cat_cols:
    df[c] = df[c].astype('category')


df = df.drop_duplicates()

df['children'] = df['children'].fillna(0).astype(int)
df['babies']   = df['babies'].fillna(0).astype(int)
df['adults']   = df['adults'].astype(int)
```
