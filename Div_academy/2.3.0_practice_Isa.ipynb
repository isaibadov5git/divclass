```python
import yfinance as yf
import datetime

data = yf.download(
    ['NVDA'], 
    start=datetime.datetime(2022,1,1),
    end=datetime.datetime.today()
    )
df = data.droplevel(level=1, axis='columns')
df.head()
df2025 = df.loc['2025', :]
df2025_rolling = df2025.rolling(window=30).mean().shift(-15)
df2025_rolling.tail(3)
plt.style.use('seaborn-v0_8-whitegrid')

plt.plot(df2025.index, df2025['Close'],linestyle = '',marker='.')
plt.plot(df2025_rolling.index, df2025_rolling['Close'])

plt.gcf().autofmt_xdate()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

x = np.linspace(0,6,50)
y = np.linspace(0,5,60)

X,Y = np.meshgrid(x,y)

def f(x, y):
    return np.sin(x) ** 10 + np.cos(10 + y * x) * np.cos(x)

Z = f(X,Y)

countour = plt.contour(X, Y, Z, levels=3, colors='black')
plt.clabel(countour, fontsize=7)
plt.imshow(Z, origin='lower', extent=[0,6,0,5], alpha=0.5, cmap='magma')
plt.colorbar()
plt.show()
```
