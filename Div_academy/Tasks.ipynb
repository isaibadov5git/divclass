```python
import seaborn as sns
import pandas as pd
import numpy as np

df = sns.load_dataset("titanic")
df.head()

df[df['sex'] == 'female']

df[df['age'] > 18 and df['sex'] == 'female']
df.loc[:,['survived','pclass','age','fare']]
```

```python
df.head(100)
df2 = df.loc[:,['age','fare','survived']]
# df2[df2['fare'] < 20] = np.nan
# df
df2
```

```python
def myfunction(age):
    if pd.isna(age):
        return 'Unknown'
    elif age < 18:
        return 'Child'
    elif age >=18 and age < 60:
        return 'Adult'
    else:
        return "Senior"
    
df2['age_group'] = df2['age'].apply(myfunction)
df2


```

```python
def function2(age):
    if pd.isna(age):
        return round(df2['age'].mean())
    else:
        return age

df2['age_filled'] = df2['age'].apply(function2)
df2
    
```
